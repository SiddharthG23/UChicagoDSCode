```python
df = pd.read_csv('synthetic.csv')
```

```python
X = df.drop('DependentVariable',axis=1)
y = df['DependentVariable']

X_train = X[:1400]
X_test = X[1400:]
y_train = y[:1400] 
y_test = y[1400:] 
```

```python
gs = GridSearchCV(KNeighborsRegressor(), 
                  param_grid = {'n_neighbors': [1,2,4,8,16,32],
                  'metric':['euclidean']}, scoring = 'r2',
                  cv=10)
gs.fit(X_train,y_train)
print(gs.best_params_)

clf = gs.best_estimator_
y_pred = clf.predict(X_test)
print(r2_score(y_test,y_pred))

gs = GridSearchCV(KNeighborsRegressor(), 
                  param_grid = {'n_neighbors': [1,2,4,8,16,32],
                  'metric':['manhattan']}, scoring = 'r2',
                  cv=10)
gs.fit(X_train,y_train)
print(gs.best_params_)

clf = gs.best_estimator_
y_pred = clf.predict(X_test)
print(r2_score(y_test,y_pred))

gs = GridSearchCV(KNeighborsRegressor(), 
                  param_grid = {'n_neighbors': [1,2,4,8,16,32],
                  'metric':['hamming']}, scoring = 'r2',
                  cv=10)
gs.fit(X_train,y_train)
print(gs.best_params_)

clf = gs.best_estimator_
y_pred = clf.predict(X_test)
print(r2_score(y_test,y_pred))
```

```python
#Question 3 
import numpy as np
import pickle 
from torch import nn, from_numpy
from torch.optim import Adam
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt
X_train, X_test, y_train, y_test = pickle.load(open('emnist.p','rb'))
```

```python
# Re-shaping the data
X_train = np.reshape(X_train, (60000,1,28,28))
X_test  = np.reshape(X_test, (10000,1,28,28))

# Converting to Torch tensors
X_train = from_numpy(X_train)
X_test = from_numpy(X_test)
y_train = from_numpy(y_train)
y_test = from_numpy(y_test)
```

```python
model = nn.Sequential(nn.Flatten(),
                      nn.Linear(28*28,500),
                      nn.ReLU(),
                      nn.Linear(500,250),
                      nn.ReLU(),
                      nn.Linear(250,26))

loss_function = nn.CrossEntropyLoss()

optimizer = Adam(model.parameters())
```

```python
batch_size = 60000

for cycle in tqdm(range(100)):
    
    for batch in range(int(60000/batch_size)):
        # ForwardProp
        y_pred = model(X_train[batch*batch_size:(batch+1)*batch_size]) 
        loss = loss_function(y_pred, y_train[batch*batch_size:(batch+1)*batch_size])

        # BackwardProp
        optimizer.zero_grad() 
        loss.backward() 
        optimizer.step() 

    # Printing in-sample and out-of-sample accuracy    
    print('In-sample cross-entropy: ' + str(loss.item()))
    y_pred = model(X_test)
    print('Out-of-sample accuracy: ' + str((y_pred.argmax(1) == y_test).sum().item() / 10000))
```
